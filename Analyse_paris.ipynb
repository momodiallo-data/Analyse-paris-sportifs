```python
print("Nombre de transactions :", len(df))
print("Total des mises :", df["mise_fcfa"].sum())
print("Total des gains versés :", df["gain_verse_fcfa"].sum())
print("Mise moyenne :", df["mise_fcfa"].mean())
```

```python
analyse_sports = df.groupby("type_pari").agg(
    nombre_paris=("type_pari", "count"),
    total_mises=("mise_fcfa", "sum"),
    total_gains_verses=("gain_verse_fcfa", "sum")
).reset_index()

analyse_sports = analyse_sports.sort_values(
    by="total_mises",
    ascending=False
)

analyse_sports
```

```python
analyse_points = df.groupby("point_vente").agg(
    nombre_transactions=("point_vente", "count"),
    total_mises=("mise_fcfa", "sum"),
    total_gains_verses=("gain_verse_fcfa", "sum")
).reset_index()

analyse_points.sort_values(by="total_mises", ascending=False)
```

```python
df["date"] = pd.to_datetime(df["date"])

analyse_mensuelle = df.groupby(
    df["date"].dt.to_period("M")
).agg(
    nombre_transactions=("date", "count"),
    total_mises=("mise_fcfa", "sum"),
    total_gains_verses=("gain_verse_fcfa", "sum")
).reset_index()

analyse_mensuelle["date"] = analyse_mensuelle["date"].astype(str)

analyse_mensuelle
```

```python
analyse_statuts = df.groupby("statut").agg(
    nombre_transactions=("statut", "count"),
    total_mises=("mise_fcfa", "sum"),
    total_gains_verses=("gain_verse_fcfa", "sum")
).reset_index()

analyse_statuts
```
